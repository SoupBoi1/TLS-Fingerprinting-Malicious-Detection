```python
daCon[label_word] = ["NORMAL" for item in daCon["Filename"]]
dmCon[label_word] =[
    str(item).split('/')[2] if len(str(item).split('/')) >= 3 else None 
    for item in dmCon["Filename"]
]
maCon[label_word] = ["NORMAL" for item in maCon["Filename"]]
mmCon[label_word] = [
    "MALWARE" if item == "M" 
    else "advertising/analytics" if item == "A" 
    else "NAN"
    for item in mmCon["Type"]
]
```
